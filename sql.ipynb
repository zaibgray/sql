```python
%%sql
CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    grade TEXT
);

INSERT INTO students (name, age, grade) VALUES
('John', 20, 'A'),
('Sarah', 22, 'B'),
('Michael', 19, 'A'),
('Emma', 21, 'C'),
('David', 23, 'B');
```

```python
%%sql
SELECT * FROM students;
```
