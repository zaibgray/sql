### What are SQL joins?

In SQL (Structured Query Language) a join is a way to combine data from two or more database tables on a related column between them.

Join are used when we want to query information that is distributed across multiple table in a database, and the information we need is not contained in a single table. By joining table together, we can create a virtual table contains all of the information we need for our query.

### Cross Join -> Cartesian Products
![alt text](images/sql_cross_join.png)

In SQL, a cross join (also known as a Cartesian product) is a type of join that returns the cartesian product of the two tables being joined. In other words. it returns all possible combinations of rows from the tables.

Cross joins are not commonly used in practice, but they can useful in certain scenarios, such as generating test data or exploring all possible combinations of items in a product catalogue. However, it's important to be cautious when using cross joins with large tables, as they can generate a very large result set, which can be resource-intensive and slow to process.
