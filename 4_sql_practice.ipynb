```python
CREATE DATABASE insurance
```

```python
CREATE TABLE users(
	id SERIAL NOT NULL,
	patient_id INTEGER NOT NULL,
	age INTEGER NOT NULL,
	gender VARCHAR(255),
	bmi DECIMAL(4,1),
	blood_pressure INTEGER,
	diabetic VARCHAR(5),
	chidren INTEGER,
	smoker VARCHAR(5),
	region VARCHAR(50),
	claim DECIMAL(10,2)
)
```

```python
ALTER TABLE users
ALTER COLUMN age DROP NOT NULL;
```

```python
-- Show records of 'male' patient from 'southwest' region.
SELECT gender, region FROM public.users
WHERE gender = 'male' AND region = 'southwest'
```

```python
-- 2. Show all records having bmi in range 30 to 45 both inclusive.
SELECT * FROM public.users
WHERE bmi >= 30 AND bmi <= 45
```

```python
-- 3. Show minimum and maximum bloodpressure 
-- of diabetic patient who smokes.
-- Make column names as MinBP and MaxBP respectively.
SELECT
    MAX(blood_pressure) AS MaxBP,
    MIN(blood_pressure) AS MinBP
FROM public.users
WHERE smoker = 'Yes'
  AND diabetic = 'Yes';
```

```python
-- 4. Find no of unique patients
-- who are not from southwest region.
SELECT COUNT(DISTINCT patient_id) AS unique_patients
FROM public.users
WHERE region != 'southwest';
```

```python
-- 5. Total claim amount from male smoker.
SELECT SUM(claim) AS total_claim 
FROM public.users
WHERE gender = 'male'
	AND smoker = 'Yes'
```

```python
-- Select all records of south region.
SELECT * FROM public.users
WHERE region LIKE 'south%'

-- % means any characters after south 
```

```python
-- 7. No of patient having normal blood pressure. 
-- Normal range[90-120]
SELECT COUNT(patient_id) AS patient_cont
FROM public.users
WHERE blood_pressure BETWEEN 90 AND 120
```

```python
-- 8. No of pateint belo 17 years of age having normal blood pressure as per below formula -
    -- - BP normal range = 80+(age in years × 2) to 100 + (age in years × 2)

    -- - Note: Formula taken just for practice, don't take in real sense.

SELECT COUNT(patient_id)
FROM public.users
WHERE age < 17
  AND blood_pressure BETWEEN 80 + (age * 2)
                         AND 100 + (age * 2);
```

```python
-- 9. What is the average claim amount for 
-- non-smoking female patients who are diabetic?

SELECT AVG(claim) AS average_claim
FROM public.users
WHERE gender = 'female'
  AND smoker = 'No'
  AND diabetic = 'Yes';
```

```python
-- 10. Write a SQL query to update the claim amount
-- for the patient with PatientID = 1234 to 5000.

UPDATE public.users
SET claim = '5000'
WHERE patient_id = '1234';
```

```python
-- 11. Write a SQL query to delete all records 
-- for patients who are smokers and have no children.
-- be careful you can't undo that so make sure you've taken backup

DELETE FROM public.users
WHERE smoker = 'Yes' 
	and chidren = '0';
```
