# SQL DML (Data Manipulation Language)
DML commands manage row-level data records inside existing tables. They handle data state changes but never touch the schema structure.

### The 4 Essential Actions
- **INSERT** (Add): Pushes new rows into a table.
- **UPDATE** (Change): Modifies existing values. Requires WHERE to avoid breaking all rows.
- **DELETE** (Remove): Erases specific rows. Requires WHERE to avoid wiping the entire table.
- **MERGE** (Upsert): Matches data between tables. Updates if a row exists; inserts if it is new.

### Critical Concepts to Remember
- **Transactional**: DML actions are temporary until saved. You can use ROLLBACK to undo mistakes before issuing a COMMIT.

- **The DELETE** vs. **TRUNCATE Trap**:
    - **DELETE** (DML) removes rows one-by-one, logs every change, and can be rolled back.
    - **TRUNCATE** (DDL) drops and recreates the whole table structure instantly. It cannot be rolled back and resets counter IDs.

# DQL (Data Query Role ): 
SELECT is technically DQL (Data Query Language), but practically grouped with DML because it reads raw row data.

# PostgreSQL & pgAdmin: CSV Import Guide

## Overview
This document outlines the steps to create a schema for `smartphones_cleaned_v6.csv` in PostgreSQL and import the dataset using the pgAdmin GUI without encountering data type or null-constraint errors.

---

## 1. Table Schema (`CREATE TABLE`)

Because the CSV file contains missing values (`NaN`) and specific data formats (floating-point numbers for specs and booleans), run the following SQL script to set up the matching structure:

```sql
DROP TABLE IF EXISTS public.smartphones;

CREATE TABLE public.smartphones (
    id SERIAL PRIMARY KEY,
    brand_name VARCHAR(50),
    model VARCHAR(150),
    price NUMERIC(12, 2),
    rating NUMERIC(4, 1),
    has_5g BOOLEAN,
    has_nfc BOOLEAN,
    has_ir_blaster BOOLEAN,
    processor_brand VARCHAR(50),
    num_cores NUMERIC(4, 1),
    processor_speed NUMERIC(4, 2),
    battery_capacity NUMERIC(8, 2),
    fast_charging_available INT,
    fast_charging NUMERIC(6, 1),
    ram_capacity NUMERIC(6, 1),
    internal_memory NUMERIC(8, 1),
    screen_size NUMERIC(4, 2),
    refresh_rate INT,
    num_rear_cameras INT,
    num_front_cameras NUMERIC(4, 1),
    os VARCHAR(50),
    primary_camera_rear NUMERIC(6, 1),
    primary_camera_front NUMERIC(6, 1),
    extended_memory_available INT,
    extended_upto NUMERIC(8, 1),
    resolution_width INT,
    resolution_height INT
);

## 2. Import Steps in pgAdmin GUI

### Step 1: Open Import Wizard
1. In pgAdmin's Object Explorer, expand **Databases** > **Schemas** > **public** > **Tables**.
2. Right-click the **`smartphones`** table and select **Import/Export Data...**

### Step 2: General Tab Configuration
* Set **Direction** to `Import`.
* Click **Filename** (`...`) and select your `smartphones_cleaned_v6.csv` file.
* Set **Format** to `csv`.

### Step 3: Options Tab Configuration
* Set **Header** to `Yes`.
* Set **Delimiter** to `,`.
* Ensure **Null Value** is left empty or set to `""` to properly parse empty cells as `NULL`.

### Step 4: Columns Tab Configuration
* Under **Columns to import**, **uncheck `id`**.
  > **Note:** The `id` column is auto-generated by PostgreSQL (`SERIAL`). Unchecking it aligns the CSV's first column (`brand_name`) with the database structure.

### Step 5: Execute
* Click **OK** to complete the import process.
