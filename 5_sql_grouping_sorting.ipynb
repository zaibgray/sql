# Sorting Data
1. find top 5 samsung phones with biggest screen size
2. sort all the phone in decending order of number of total cameras
3. sort data on the basis of ppi in decreasing order
4. find the phone with 2nd largest battery
5. find the name and rating of the worst rated apple phone
6. sort phones alphabetically and then on the basic of rating in desc order
7. sort phones alphabetically and then on the basic of price in asc order

```python
SELECT * FROM public.smartphones
```

```python
-- 1. find top 5 samsung phones with biggest screen size
SELECT model, screen_size FROM public.smartphones
WHERE brand_name = 'samsung'
ORDER BY screen_size DESC LIMIT 5

-- ORDER BY -> sort
-- LIMIT as head()
```

```python
-- 2. sort all the phone in decending order of number of total cameras
SELECT 
    model, 
    COALESCE(num_front_cameras, 0) + COALESCE(num_rear_cameras, 0) AS total_cameras 
FROM public.smartphones 
ORDER BY total_cameras DESC;
```

```python
-- 3. sort data on the basis of ppi in decreasing order
SELECT model,
ROUND(SQRT(resolution_width * resolution_width + resolution_height * resolution_height)/screen_size) AS ppi
FROM public.smartphones 
ORDER BY ppi DESC;
```

```python
-- 4. find the phone with 2nd largest battery
SELECT model, battery_capacity
FROM public.smartphones
WHERE battery_capacity IS NOT NULL
ORDER BY battery_capacity DESC
LIMIT 1 OFFSET 1;
```

```python
SELECT model, battery_capacity
FROM public.smartphones
WHERE battery_capacity IS NOT NULL
ORDER BY battery_capacity DESC, model ASC
LIMIT 1 OFFSET 5;
```

```python
-- 4. find the phone with 2nd lowest battery
SELECT model, battery_capacity
FROM public.smartphones
WHERE battery_capacity IS NOT NULL
ORDER BY battery_capacity ASC, model ASC
LIMIT 1 OFFSET 1;
```

```python
-- 5. find the name and rating of the worst rated apple phone
SELECT model, rating
FROM public.smartphones
WHERE brand_name = 'apple' AND rating IS NOT NULL
ORDER BY rating ASC LIMIT 1
```

```python
-- 6. sort phones alphabetically and then on the basic of rating in desc order
SELECT *
FROM public.smartphones
WHERE brand_name IS NOT NULL AND price IS NOT NULL
ORDER BY brand_name ASC , price ASC
```

```python
SELECT *
FROM public.smartphones
WHERE brand_name IS NOT NULL AND rating IS NOT NULL
ORDER BY brand_name ASC , rating DESC
```

# Grouping Data
1. Group smartphones by brand and get the count, average price, max rating, avg screen size and avg battery capacity
--> Group By Animation
2. Group smartphones by whether they have an NFC and get the average price and rating
3. Group smartphones by the extended memory available and get the average price
4. Group smartphones by the brand and processor brand and get the count of models and the average primary camera resolution (rear)
5. find top 5 most costly phone brands
6. which brand makes the smallest screen smartphones
7. Avg price of 5g phones vs avg price of non 5g phones
8. Group smartphones by the brand, and find the brand with the highest number of models that have both NFC and an IR blaster
9. Find all samsung 5g enabled smartphones and find out the avg price for NFC and Non-NFC phones
10. find the phone name, price of the costliest phone

```python
-- 1. Group smartphones by brand and get the count, average price, 
-- max rating, avg screen size and avg battery capacity

SELECT brand_name, COUNT(*) AS num_phones,
ROUND(AVG(price), 2) AS avg_price,
MAX(rating) AS max_rating,
ROUND(AVG(screen_size), 2) AS avg_screen_size,
ROUND(AVG(battery_capacity),2) AS avg_battery_capacity
FROM public.smartphones
GROUP BY brand_name
ORDER BY num_phones DESC 
LIMIT 15
```

```python
-- Group smartphones by whether they have an NFC
--  and get the average price and rating

SELECT has_nfc, 
AVG(price) AS avg_price,
AVG(rating) AS rating
FROM public.smartphones
GROUP by has_nfc
```

```python
-- 3. Group smartphones by the extended memory 
-- available and get the average price
SELECT extended_memory_available, 
AVG(price) AS avg_price,
AVG(rating) AS rating
FROM public.smartphones
GROUP by extended_memory_available
```

```python
-- 4. Group smartphones by the brand and processor brand and get the count of models
--  and the average primary camera resolution (rear)
SELECT brand_name,
processor_brand, os,
COUNT(*) AS num_phones,
ROUND(AVG(primary_camera_rear)) AS avg_camera_resolution
FROM public.smartphones
GROUP BY brand_name, processor_brand, os
ORDER BY brand_name ASC 
```

```python
-- 5. find top 5 most costly phone brands
SELECT brand_name,
ROUND(AVG(price), 2) AS avg_price
FROM public.smartphones
GROUP BY brand_name
ORDER BY avg_price DESC 
LIMIT 5
```

```python
-- 6. which brand makes the smallest screen smartphones
SELECT brand_name,
ROUND(AVG(screen_size), 2) AS avg_screen_size
FROM public.smartphones
GROUP BY brand_name
ORDER BY avg_screen_size ASC
LIMIT 1
```

```python
-- 7. Avg price of 5g phones vs avg price of non 5g phones
SELECT has_5g, 
AVG(price) AS avg_price,
AVG(rating) AS rating
FROM public.smartphones
GROUP by has_5g
```

```python
-- 8. Group smartphones by the brand, and find the brand with the highest number
--  of models that have both NFC and an IR blaster
SELECT brand_name,
COUNT(*) AS total
FROM public.smartphones
WHERE has_nfc = True AND has_ir_blaster = TRUE
GROUP BY brand_name
ORDER BY total DESC LIMIT 1
```

```python
-- 9. Find all samsung 5g enabled smartphones
--  and find out the avg price for NFC and Non-NFC phones
SELECT has_nfc,
AVG(price) AS avg_price
FROM public.smartphones
WHERE brand_name = 'samsung'
GROUP BY has_nfc
```

```python
-- 10. find the phone name, price of the costliest phone
SELECT model, price
FROM public.smartphones
ORDER BY price DESC LIMIT 1

```

# Having clause
1. find the avg rating of smartphone brands which have more than 20 phones
2. Find the top 3 brands with the highest avg ram that have a refresh rate of at least 90 Hz and fast charging available and dont consider brands which have less than 10 phones
3. find the avg price of all the phone brands with avg rating > 70 and num phones more than 10 among all 59 enabled phones

```python
SELECT brand_name,
       COUNT(*) AS total,
       ROUND(AVG(rating)) AS avg_rating
FROM public.smartphones
GROUP BY brand_name
HAVING COUNT(*) > 40
ORDER BY avg_rating DESC;
```

```python
SELECT brand_name,
AVG(ram_capacity) AS avg_ram
FROM public.smartphones
WHERE refresh_rate > 90 AND fast_charging_available = 1
GROUP BY brand_name
HAVING COUNT(*) > 10
ORDER BY avg_ram DESC LIMIT 3
```

```python
SELECT brand_name,
AVG(price) AS avg_price
FROM public.smartphones
WHERE has_5g  = True
GROUP BY brand_name
HAVING AVG(rating) > 70 AND COUNT(*) > 10
```

# Practice
- find the top 5 batsman in IPL
- find the 2nd highest 6 hitter in IPL
- find the top 5 batsman with highest strike rate who have played a min of 1000 balls

```python
CREATE TABLE ipl (
    id BIGINT,
    innings INT,
    overs INT,
    ballnumber INT,
    batter VARCHAR(255),
    bowler VARCHAR(255),
    non_striker VARCHAR(255), 
    extra_type VARCHAR(50), 
    batsman_run INT,
    extras_run INT,
    total_run INT,
    non_boundary INT,
    is_wicket_delivery INT,
    player_out VARCHAR(255),
    kind VARCHAR(100),
    fielders_involved VARCHAR(255),
    batting_team VARCHAR(255)
);
```

```python
-- find the top 5 batsman in IPL
SELECT 
    batter, 
    SUM(total_run) AS total_run
FROM public.ipl
GROUP BY batter
ORDER BY total_run DESC LIMIT 5;
```

```python
-- - find the 2nd highest 6 hitter in IPL
SELECT 
	batter,
	COUNT(batsman_run) AS total_six
FROM public.ipl
WHERE batsman_run = 6
GROUP BY batter
ORDER BY total_six DESC 
	LIMIT 1 OFFSET 1;
```

```python
SELECT batter, id, SUM(batsman_run) as score
FROM public.ipl
GROUP BY batter, id
HAVING SUM(batsman_run) >= 100
ORDER BY batter DESC
```

```python
-- - find the top 5 batsman with highest strike rate who have played a min of 1000 balls
SELECT
	batter, 
	SUM(batsman_run), 
	COUNT(batsman_run),
	ROUND((SUM(batsman_run) * 100.0 / COUNT(batsman_run)), 2) AS sr
FROM public.ipl
GROUP BY batter
HAVING COUNT(batsman_run) > 1000 
ORDER BY sr DESC LIMIT 5
```
